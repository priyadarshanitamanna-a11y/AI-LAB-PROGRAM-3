Problem 3: Checking Routes Between Cities Using DFS
Problem Statement:
A small road network connects several cities. Write a Python program using DFS that:
1. Checks whether a route exists between two given cities.
2. Finds all separate groups of cities that are connected to each other but not to any
other group (these are called connected components).

```python
from collections import deque

# Capacities of the two jugs
CAP_A = 4
CAP_B = 3

# BFS queue: (state, path)
# State = (amount_in_A, amount_in_B)
queue = deque([((0, 0), [])])

# Keep track of visited states
visited = set()
visited.add((0, 0))

while queue:
    (a, b), path = queue.popleft()

    # Goal: exactly 2 litres in either jug
    if a == 2 or b == 2:
        print("Minimum steps:", len(path))
        print("\nSequence of operations:")

        for i, operation in enumerate(path, 1):
            print(f"{i}. {operation}")

        print(f"\nFinal state: ({a}, {b})")
        break

    # Generate all possible next states

    next_states = []

    # 1. Fill Jug A
    next_states.append(((CAP_A, b), "Fill Jug A"))

    # 2. Fill Jug B
    next_states.append(((a, CAP_B), "Fill Jug B"))

    # 3. Empty Jug A
    next_states.append(((0, b), "Empty Jug A"))

    # 4. Empty Jug B
    next_states.append(((a, 0), "Empty Jug B"))

    # 5. Pour A -> B
    amount = min(a, CAP_B - b)
    next_states.append(((a - amount, b + amount), "Pour A -> B"))

    # 6. Pour B -> A
    amount = min(b, CAP_A - a)
    next_states.append(((a + amount, b - amount), "Pour B -> A"))

    # Add unvisited states to the queue
    for new_state, operation in next_states:
        if new_state not in visited:
            visited.add(new_state)
            queue.append((new_state, path + [operation]))
```
